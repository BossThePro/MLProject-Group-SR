```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

df = pd.read_csv("results/results_scratch_test.csv")
df.describe()
# df[df["ClaimNb"] == 0].describe()
df[df["ClaimNb"] > 0].describe()
# df[df["ClaimNb"] == 2].describe()
# df[df["ClaimNb"] == 3].describe()
```

```python
import sklearn.metrics

y_true = np.array(df.loc[:,"ClaimNb"])
y_pred = np.array(df.loc[:,"Predicted"])

print(y_true)
print(y_pred)

print(sklearn.metrics.mean_poisson_deviance(y_true,y_pred))
print(sklearn.metrics.r2_score(y_true, y_pred))
print(sklearn.metrics.d2_tweedie_score(y_true, y_pred,power=1))
```
