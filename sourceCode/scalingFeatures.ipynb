# This jupyter notebook file aims to scale our features in accordance with standard practices, i.e. fitting a scaler to the training data and applying that fitted scaler on both training and test data

```python
import pandas as pd 
from sklearn.preprocessing import StandardScaler
```

```python
# Loading datasets
df_train = pd.read_csv("../data/claims_train_final.csv")
df_test = pd.read_csv("../data/claims_test_final.csv")
df_train
```

```python
df_test
```

```python
# We need to scale each of the four following features: VehPower, VehAge, DrivAge, BonusMalus (and maybe density if we keep it, will do it for now and might remove later)

# VehPower
scalerVehPower = StandardScaler() 
trainingVehPower = df_train[["VehPower"]]
scalerVehPower.fit(trainingVehPower)
df_train["VehPower"] = scalerVehPower.transform(trainingVehPower)
testVehPower = df_test[["VehPower"]]
df_test["VehPower"] = scalerVehPower.transform(testVehPower)


print(df_train["VehPower"].describe().round(2))
print(df_test["VehPower"].describe().round(2))

# VehAge
scalerVehAge = StandardScaler() 
trainingVehAge = df_train[["VehAge"]]
scalerVehAge.fit(trainingVehAge)
df_train["VehAge"] = scalerVehAge.transform(trainingVehAge)
testVehAge = df_test[["VehAge"]]
df_test["VehAge"] = scalerVehAge.transform(testVehAge)


print(df_train["VehAge"].describe().round(2))
print(df_test["VehAge"].describe().round(2))


# DrivAge
scalerDrivAge = StandardScaler() 
trainingDrivAge = df_train[["DrivAge"]]
scalerDrivAge.fit(trainingDrivAge)
df_train["DrivAge"] = scalerDrivAge.transform(trainingDrivAge)
testDrivAge = df_test[["DrivAge"]]
df_test["DrivAge"] = scalerDrivAge.transform(testDrivAge)


print(df_train["DrivAge"].describe().round(2))
print(df_test["DrivAge"].describe().round(2))


# BonusMalus
scalerBonusMalus = StandardScaler() 
trainingBonusMalus = df_train[["BonusMalus"]]
scalerBonusMalus.fit(trainingBonusMalus)
df_train["BonusMalus"] = scalerBonusMalus.transform(trainingBonusMalus)
testBonusMalus = df_test[["BonusMalus"]]
df_test["BonusMalus"] = scalerBonusMalus.transform(testBonusMalus)


print(df_train["BonusMalus"].describe().round(2))
print(df_test["BonusMalus"].describe().round(2))

# As expected we see vehpower with the same min and max due to them not being cleaned and having the same range, while we see different min/max values for the other three features


# Density (for now)
scalerDensity = StandardScaler() 
trainingDensity = df_train[["Density"]]
scalerDensity.fit(trainingDensity)
df_train["Density"] = scalerDensity.transform(trainingDensity)
testDensity = df_test[["Density"]]
df_test["Density"] = scalerDensity.transform(testDensity)


print(df_train["Density"].describe().round(2))
print(df_test["Density"].describe().round(2))
```
