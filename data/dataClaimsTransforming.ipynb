# Here we include the actual data cleaning part for our training data, such that it is easily accessible

## All graphs for everything can be found in dataCleaning.ipynb and test.ipynb

Loading Libraries and dataset

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
df = pd.read_csv("claims_train.csv")
df
```

Removal of IDPol feature

```python
df = df.drop(columns=["IDpol"])
```

Removing Exposure values above 1

```python
exposureMask = df["Exposure"] <= 1
df = df[exposureMask]
df
```

Removing Vehicle Age values above 40

```python
vehicleAgeMask = df["VehAge"] <= 40
df = df[vehicleAgeMask]
df
```

Removing Driver Age Values above 92


```python
driverAgeMask = df["DrivAge"] <= 92
df = df[driverAgeMask]
df
```

Removing bonus-malus scores greater than 160

```python
bonusMalusMask = df["BonusMalus"] <= 160
df = df[bonusMalusMask]
df
```

Export final dataset

```python
df.to_csv("claims_train_final.csv", index=False)
```

### For the test dataset we will only be removing IDpol as it is a useless measure, we will not be clamping values or removing outliers as we are treating this dataset as real observed data where we would not want to remove data points at will.

```python
df_test = pd.read_csv("claims_test.csv")
df_test = df_test.drop(columns=["IDpol"])
df_test.to_csv("claims_test_final.csv", index=False)
```
